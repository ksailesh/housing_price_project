# Preprocessing: Missing values
1.   Fill the missing values
  *    Imputation with Median value created 2 invalid records found in Outlier Analysis
2.   Remove samples with missing values


```python
# Drop 189 records (~1%) without total_bedrooms data.
# Convert Float type to INT type

df_nonnull = df.dropna(subset=['total_bedrooms'], inplace=False)
df_nonnull.loc[:, "total_bedrooms"] = df_nonnull["total_bedrooms"].astype('Int64')
df_nonnull.describe()
```

```python
# Check the missing data in the dataframe

# Check for missing data
missing_data = df.isnull().sum()
missing_data

## 189 samples do not have total_bedrooms data
```

```python
# Impute the missing total_bedrooms data using SimplyImputer

from sklearn.impute import SimpleImputer

# Create a SimpleImputer object with strategy='median'
imputer = SimpleImputer(strategy='median')

# Fit the imputer on the 'total_bedrooms' column
imputer.fit(df[['total_bedrooms']])

# Transform the 'total_bedrooms' column to impute missing values
df_imputed = df.copy()  # Create a copy to avoid modifying the original DataFrame
df_imputed['total_bedrooms'] = imputer.transform(df[['total_bedrooms']])
df_imputed["total_bedrooms"] = df_imputed["total_bedrooms"].astype('Int64')

# Now df_imputed contains the imputed 'total_bedrooms' column
print(df_imputed.isnull().sum())
df_imputed.dtypes
```

```python
# Find the distribution of the each data attribute

import matplotlib.pyplot as plt
import seaborn as sns

# Function to detect outliers using IQR
def find_outliers_iqr(data):
    q1 = data.quantile(0.25)
    q3 = data.quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = data[(data < lower_bound) | (data > upper_bound)]
    return outliers

# Iterate through numerical columns and find outliers
for column in df[['housing_median_age', 'total_rooms', 'total_bedrooms', 'population', 'households', 'median_income', 'median_house_value']]:
    outliers = find_outliers_iqr(df[column])
    print(f"Outliers in {column}:")
    print(outliers)

    # Visualization
    plt.figure(figsize=(8, 6))
    sns.boxplot(x=df[column])
    plt.title(f'Boxplot of {column} with Outliers')
    plt.show()

    # Histogram for distribution visualization
    plt.figure(figsize=(8, 6))
    plt.hist(df[column], bins=30)
    plt.title(f'Histogram of {column} with Outliers')
    plt.xlabel(column)
    plt.ylabel('Frequency')
    plt.show()
```

# Outlier Analysis

image.png

```python
# Enrich with new features

df_enriched = df_imputed.copy()

df_enriched['rooms_per_household'] = df_enriched['total_rooms'] / df_enriched['households']
df_enriched['population_per_household'] = df_enriched['population'] / df_enriched['households']
df_enriched['bedrooms_per_household'] = df_enriched['total_bedrooms'] / df_enriched['households']
df_enriched['bedrooms_per_room'] = df_enriched['total_bedrooms'] / df_enriched['total_rooms']
df_enriched['income_per_population'] = df_enriched['median_income'] / df_enriched['population']
df_enriched['income_per_household'] = df_enriched['median_income'] / df_enriched['households']

df_enriched.head()
```

```python
# Visualize the enriched columns against median housing value and ocean proximity
import seaborn as sns

for feature in ['rooms_per_household', 'population_per_household', 'bedrooms_per_household',
                'bedrooms_per_room', 'income_per_population', 'income_per_household']:
    plt.figure(figsize=(10, 6))
    sns.scatterplot(x='median_house_value', y=feature, hue='ocean_proximity', data=df_enriched)
    plt.title(f'{feature} vs. Median House Value by Ocean Proximity')
    plt.show()
```

```python
# Create a copy of the DataFrame
df_outlier = df_enriched.copy()

# Initialize the 'outlier_flag' column with 0
df_outlier['rooms_per_household_flag'] = 0
df_outlier['population_per_household_flag'] = 0
df_outlier['bedrooms_per_household_flag'] = 0
df_outlier['bedrooms_per_room_flag'] = 0
df_outlier['income_per_population_flag'] = 0
df_outlier['income_per_household_flag'] = 0
df_outlier['total_outlier_flag'] = 0

# Assign outlier flags based on the specified conditions
df_outlier.loc[df_outlier['rooms_per_household'] > 100, 'rooms_per_household_flag'] = 1
df_outlier.loc[df_outlier['population_per_household'] > 200, 'population_per_household_flag'] = 1
df_outlier.loc[df_outlier['bedrooms_per_household'] > 20, 'bedrooms_per_household_flag'] = 1
df_outlier.loc[df_outlier['bedrooms_per_room'] > 1.5, 'bedrooms_per_room_flag'] = 1
df_outlier.loc[df_outlier['income_per_population'] > 0.6, 'income_per_population_flag'] = 1
df_outlier.loc[df_outlier['income_per_household'] > 1.5, 'income_per_household_flag'] = 1

# Select the columns representing individual outlier flags
outlier_flag_columns = ['rooms_per_household_flag', 'population_per_household_flag',
                        'bedrooms_per_household_flag', 'bedrooms_per_room_flag',
                        'income_per_population_flag', 'income_per_household_flag']

# Calculate the total outlier flag by summing the individual flags
df_outlier['total_outlier_flag'] = df_outlier[outlier_flag_columns].sum(axis=1)

df_outlier[df_outlier['total_outlier_flag']!=0]
# df[df_outlier['total_outlier_flag']!=0]
```

# Save Cleansed and Processed Datasets in Google Drive

```python
import pandas as pd

# Exclude missing value records, Outlier records and flag columns
df_filtered = df_outlier[df_outlier['total_outlier_flag'] == 0]
flag_columns = [col for col in df_filtered.columns if '_flag' in col]
df_no_flags = df_filtered.drop(columns=flag_columns).dropna(subset=['total_bedrooms'], inplace=False)

# One-hot encode the 'ocean_proximity' column
df_enriched_features = pd.get_dummies(df_no_flags, columns=['ocean_proximity'], drop_first=True)
display(df_enriched_features.head())
df.to_pickle("df_enriched_features.pkl")

df_without_enriched_features = df_enriched_features.drop(columns=['rooms_per_household', 'population_per_household', 'bedrooms_per_household', 'bedrooms_per_room', 'income_per_population', 'income_per_household'])
display(df_without_enriched_features.head())
df.to_pickle("df_without_enriched_features.pkl")
```
